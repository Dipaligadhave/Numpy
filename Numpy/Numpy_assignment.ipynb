````python
#**Q2.** Simulate a **train/validation/test split** (70/15/15) on 500 samples:
rng = np.random.default_rng(0)
samples=500
indices=np.arange(samples)

#-Shuffle indices with `rng.permutation`.
rng.shuffle(indices)

#- Slice into three groups.
train_end =int(0.70* samples)
val_end=train_end+int(0.15*samples)

train_idx=indices[:train_end]
val_idx=indices[train_end:val_end]
test_idx=indices[val_end:]

print("Train:", len(train_idx))
print("Validation:", len(val_idx))
print("Test:", len(test_idx))
#- Confirm no indices overlap between splits.
print(
    set(train_idx).isdisjoint(val_idx),
    set(train_idx).isdisjoint(test_idx),
    set(val_idx).isdisjoint(test_idx)
)
````

````python
#**Q3.** Implement a **Monte Carlo estimate of π**:
rng = np.random.default_rng(0)
N = 1_000_000

#- Generate N=1,000,000 random (x, y) points in the unit square using `rng.random`.
x = rng.random(N)
y = rng.random(N)


#- Count how many fall inside the unit circle: $x^2 + y^2 \leq 1$.
inside = np.sum(x**2 + y**2 <= 1)

#- Estimate π as `4 * inside / N`. How close is it to `np.pi`?
pi_estimate = 4 * inside / N

print("Estimated π :", pi_estimate)
print("Actual π    :", np.pi)
print("Difference  :", abs(pi_estimate - np.pi))
````
